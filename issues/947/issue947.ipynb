```python
import numpy as np
import matplotlib.pyplot as plt
import pygimli as pg
from pygimli.physics.ves import VESRhoModelling, VESModelling
from pygimli.frameworks import MultiFrameModelling
```

```python
ab2 = np.logspace(0, 2, 11)
mn2 = np.ones_like(ab2) * 0.5
# mn2 = np.array([0.5, 0.5, 0.5, 0.5, 0.5, 0.5]) / 2
# rhoa_2021 = np.array([3.12, 3.45, 3.28, 3.67, 2.94, 2.41]) #---not real data
# rhoa_2022 = np.array([2.38, 3.09, 3.56, 3.18, 2.73, 1.84]) #---not real data
fRho = VESModelling(ab2=ab2, mn2=mn2)
rhoa_2021 = fRho([2, 2, 10, 100, 30])
rhoa_2022 = fRho([2, 2, 15, 90, 35])
data = [pg.Vector(rhoa_2021), pg.Vector(rhoa_2022)]
thk = np.logspace(-0.6, -0.4, 10)
fop = MultiFrameModelling(modellingOperator=VESRhoModelling, thk=thk, ab2=ab2, mn2=mn2)
fop.setData(data)
print(fop.parameterCount)
fop.prepareJacobian()
print(fop.jacobian(), fop.jac)
```

```python
startVector = pg.Vector(fop.parameterCount, 30.0)
print(fop(startVector))
```

```python
fop.createJacobian(startVector)
print(fop.jacobian().shape, fop.jacobian())
```

```python
inv = pg.Inversion(fop=fop, verbose=True, debug=True)
inv.dataTrans = 'log' # pg.trans.TransLog()
inv.modelTrans = 'log'#1-1000' # pg.trans.TransLogLU(1, 1000)
print(inv.modelTrans)
rhoa_all = np.hstack(data)
model = inv.run(rhoa_all, relativeError=0.03, startModel=startVector, lam=30, verbose=True)
print("Final RRMS:", inv.relrms())
```
